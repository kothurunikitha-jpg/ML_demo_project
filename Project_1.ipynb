### data preprocessing

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
```

```python
x=df.drop(columns=["target_default_risk","customer_id"])
y=df["target_default_risk"]
```

```python
df["education"].value_counts()
```

```python
if "education" in x.columns:
    x["education"]=x["education"].replace({
        "Bachelors":"Bachelors",
        "HS":"High School",
        "Masters":"Masters",
        "Other":"Others",
        "PhD":"PhD",
        "Bachlors":"Bachelors"
    })
```

```python
x
```

```python
if {"income", "num_dependents"}.issubset(x.columns):
    x["income_per_dependent"] = (
        x["income"] / (x["num_dependents"] + 1)
    )
```

```python
num = x.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

cat = x.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Numeric:", num)
print("Categorical:", cat)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
def outliers(train, test, columns):
    train = train.copy()
    test = test.copy()

    for col in columns:
        Q1 = train[col].quantile(0.25)
        Q3 = train[col].quantile(0.75)

        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        train[col] = train[col].clip(lower, upper)
        test[col] = test[col].clip(lower, upper)

    return train, test


x_train, x_test = outliers(
    x_train,
    x_test,
    num
)
```

```python
# encoding,imputation,scaling
```

```python
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])
```

```python
num_pipeline
```

```python
cat_pipeline
```

```python
preprocessor = ColumnTransformer([
    ("numeric", num_pipeline, num),
    ("categorical", cat_pipeline, cat)
])
```

```python
x_train_p = preprocessor.fit_transform(x_train)
x_test_p= preprocessor.transform(x_test)
print("Original training shape:", x_train.shape)
print("Processed training shape:", x_train_p.shape)
```

```python
from sklearn.model_selection import StratifiedKFold,cross_val_score
```

```python
rfc=RandomForestClassifier()
```

```python
cv=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
scores=cross_val_score(rfc,x_train_p,y_train,scoring="f1",cv=cv)
print("f1",scores)
print("mean score:",scores.mean())
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid={
    "n_estimators": [100, 200],
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2],
    "max_features": ["sqrt", "log2"]
}
```

```python
grid=GridSearchCV(estimator=rfc,param_grid=param_grid,scoring='f1',cv=cv,n_jobs=-1,verbose=1)
grid.fit(x_train_p,y_train)
```

```python
print("best parameter:",grid.best_params_)
print("best f1 score:",grid.best_score_)
```

```python
best_rfc=grid.best_estimator_
```

```python
y_pred_rfc = grid.best_estimator_.predict(x_test_p)
```
